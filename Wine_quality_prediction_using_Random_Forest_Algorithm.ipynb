Import the required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Reading of the file of given dataset

```python
data = pd.read_csv(r"/content/winequality-red.csv")
```

Visualization of top 10 records inside dataset

```python
data.head(10)
```

Classes of target value - quality

```python
set(data.quality) # how much number it have in quality dataset
```

Description of the given dataset

```python
data.describe()
```

To check missing values

```python
data.isnull().sum() # default function in python : isnull() and sum() helps to know how many null values are present in dataset
```

Correlation matrix among all features inside the dataset

```python
plt.figure(figsize = (30,20)) #maximize the figure size
corr = data.corr() # corr() : function used to create correlation matrix
sns.heatmap(corr,annot = True)
plt.savefig("Correlation Matrix.png")
plt.show()
```

Detection of outliers using z-score

```python
from scipy import stats
z = np.abs(stats.zscore(data))
print(z)
```

z >3 (Threshold) -> this is an outlier

```python
print(np.where(z>3))
```

Remove those reports whose zscore value is greater than 3 (outlier)

```python
new_data = data[(z < 3).all(axis = 1)]
```

```python
new_data.shape
```

```python
data.shape
```

Data splitting into features(x) and target(y)

```python
from sklearn.model_selection import train_test_split
x = new_data.drop(columns = 'quality') # x - input features
y = new_data['quality'] # y - outcome
```

```python
x.head()
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2)
```

Data Modelling - Random Forest classifier

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators = 100)
model.fit(x_train,y_train)
```

prediction of data using random forest model

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

Data Evaluation on metrics

```python
from sklearn import metrics
print("Accuracy score",metrics.accuracy_score(y_test,y_pred))
```

Accuracy score : 71.4 %

```python
from sklearn import tree
tree.plot_tree(model.estimators_[0],filled = True)
plt.savefig("tree1.png")
```

```python
plt.figure(figsize = (20,20))
for i in range(len(model.estimators_)):
  tree.plot_tree(model.estimators_[i],filled = True,feature_names = x.columns)
plt.savefig("tree2.png")
```
